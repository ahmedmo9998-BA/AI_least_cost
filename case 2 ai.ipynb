#BFS ALGROTHIM

```python
import time

def bfs_measured(graph, start, goal):
    visited = set()
    queue = [[start]]

    # Trackers for Time and Space complexity
    operations = 0      # Tracks Time Complexity (how many loops)
    max_queue_size = 0  # Tracks Space Complexity (how much memory used)

    start_time = time.time()

    while queue:
        operations += 1

        # Check if the queue is the largest it has ever been
        if len(queue) > max_queue_size:
            max_queue_size = len(queue)

        path = queue.pop(0)
        current_node = path[-1]

        if current_node in visited:
            continue

        visited.add(current_node)

        if current_node == goal:
            end_time = time.time()
            return path, operations, max_queue_size, (end_time - start_time)

        adjacent_nodes = graph.get(current_node, [])

        for node in adjacent_nodes:
            new_path = path.copy()
            new_path.append(node)
            queue.append(new_path)

    end_time = time.time()
    return None, operations, max_queue_size, (end_time - start_time)
```

```python

# --- Run the measurement ---
path, ops, max_q, duration = bfs_measured(graph, start='Arad', goal='Bucharest')

print(f"Shortest path: {path}")
print("-" * 30)
print(f"Time Complexity Proxy (Operations): {ops} steps")
print(f"Space Complexity Proxy (Max Queue Size): {max_q} paths held in memory")
print(f"Actual Execution Time: {duration:.6f} seconds")
```

#uniform cost algrothim

```python
import time

def path_cost(path):
    total_cost = 0
    for (node, cost) in path:
        total_cost += cost
    return total_cost

def ucs_measured(graph, start, goal):
    visited = set()
    queue = [[(start, 0)]]

    operations = 0
    max_queue_size = 0

    start_time = time.time()

    while queue:
        operations += 1

        if len(queue) > max_queue_size:
            max_queue_size = len(queue)

        queue.sort(key=path_cost)

        path = queue.pop(0)
        node = path[-1][0]

        if node in visited:
            continue

        visited.add(node)

        if node == goal:
            end_time = time.time()
            return path, path_cost(path), operations, max_queue_size, (end_time - start_time)

        adjacent_nodes = graph.get(node, {})
        for node2, cost in adjacent_nodes.items(): # Corrected line
            new_path = path.copy()
            new_path.append((node2, cost))
            queue.append(new_path)

    end_time = time.time()
    return None, 0, operations, max_queue_size, (end_time - start_time)

    # --- Run the measurement using the name 'graph' ---
raw_path, total_cost, ops, max_q, duration = ucs_measured(graph, start='Arad', goal='Bucharest')

# Clean up the output so it just shows city names
clean_path = [city for city, cost in raw_path]

print(f"Shortest path: {clean_path}")
print(f"Total Optimal Cost: {total_cost}")
print("-" * 30)
print(f"Time Complexity Proxy (Operations): {ops} steps")
print(f"Space Complexity Proxy (Max Queue Size): {max_q} paths held in memory")
print(f"Actual Execution Time: {duration:.6f} seconds")
```

```python
import heapq

# Define the graph (moved from cell A_MIq9bUNBH6 for self-containment)
graph = {
    'Oradea': {'Zerind': 71, 'Sibiu': 151},
    'Zerind': {'Oradea': 71, 'Arad': 75},
    'Arad': {'Zerind': 75, 'Sibiu': 140, 'Timisoara': 118},
    'Timisoara': {'Arad': 118, 'Lugoj': 111},
    'Lugoj': {'Timisoara': 111, 'Mehadia': 70},
    'Mehadia': {'Lugoj': 70, 'Drobeta': 75},
    'Drobeta': {'Mehadia': 75, 'Craiova': 120},
    'Craiova': {'Drobeta': 120, 'Rimnicu Vilcea': 146, 'Pitesti': 138},
    'Rimnicu Vilcea': {'Sibiu': 80, 'Craiova': 146, 'Pitesti': 97},
    'Sibiu': {'Oradea': 151, 'Arad': 140, 'Rimnicu Vilcea': 80, 'Fagaras': 99},
    'Fagaras': {'Sibiu': 99, 'Bucharest': 211},
    'Pitesti': {'Rimnicu Vilcea': 97, 'Craiova': 138, 'Bucharest': 101},
    'Bucharest': {'Fagaras': 211, 'Pitesti': 101, 'Giurgiu': 90, 'Urziceni': 85},
    'Giurgiu': {'Bucharest': 90},
    'Urziceni': {'Bucharest': 85, 'Hirsova': 98, 'Vaslui': 142},
    'Hirsova': {'Urziceni': 98, 'Eforie': 86},
    'Eforie': {'Hirsova': 86},
    'Vaslui': {'Urziceni': 142, 'Iasi': 92},
    'Iasi': {'Vaslui': 92, 'Neamt': 87},
    'Neamt': {'Iasi': 87}
}

# Define the heuristic values (straight-line distance to Bucharest)
# These are placeholder values for demonstration; actual values would be calculated.
heuristic = {
    'Arad': 366, 'Bucharest': 0, 'Craiova': 160, 'Drobeta': 242, 'Eforie': 161,
    'Fagaras': 178, 'Giurgiu': 77, 'Hirsova': 151, 'Iasi': 226, 'Lugoj': 244,
    'Mehadia': 241, 'Neamt': 234, 'Oradea': 380, 'Pitesti': 98, 'Rimnicu Vilcea': 193,
    'Sibiu': 253, 'Timisoara': 329, 'Urziceni': 80, 'Vaslui': 199, 'Zerind': 374
}

def astar(start, goal):
    # Priority queue: (f_score, city)
    open_set = [(heuristic[start], start)]
    came_from = {}
    g_score = {city: float('inf') for city in graph}
    g_score[start] = 0

    while open_set:
        _, current = heapq.heappop(open_set)

        if current == goal:
            # Reconstruct path
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]
            path.append(start)
            path.reverse()
            return path, g_score[goal]

        for neighbor, cost in graph[current].items():
            tentative_g = g_score[current] + cost
            if tentative_g < g_score[neighbor]:
                came_from[neighbor] = current
                g_score[neighbor] = tentative_g
                f_score = tentative_g + heuristic[neighbor]
                heapq.heappush(open_set, (f_score, neighbor))

    return None, float('inf')  # No path found


# Example
path, total_cost = astar('Arad', 'Bucharest')
print("Path:", " → ".join(path))
print("Total distance:", total_cost, "km")
```
